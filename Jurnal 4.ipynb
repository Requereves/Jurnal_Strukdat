*Deskripsi:*  
Buatlah method remove_duplicates(self) yang menghapus semua node yang memiliki nilai *duplikat* (sama), hanya menyisakan kemunculan *pertama* dari setiap nilai. Linked list tidak dijamin terurut.

*Contoh:*

Input : 1 -> 3 -> 2 -> 3 -> 5 -> 1 -> 2 -> None
Output: 1 -> 3 -> 2 -> 5 -> None


*Petunjuk:*  
Gunakan sebuah set Python untuk menyimpan nilai yang sudah pernah ditemui. Lakukan traversal; jika nilai node sudah ada di set, hapus node tersebut menggunakan pointer prev. Jika belum ada, tambahkan ke set dan lanjutkan.

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class LinkedList:
    def __init__(self):
        self.head = None

    def insert_at_end(self, data):
        new_node = Node(data)
        if self.head is None:
            self.head = new_node
            return
        current = self.head
        while current.next is not None:
            current = current.next
        current.next = new_node

    def display(self):
        if self.head is None:
            print("Linked list kosong")
            return
        current = self.head
        result = []
        while current is not None:
            result.append(str(current.data))
            current = current.next
        print(" -> ".join(result) + " -> None")

    def remove_duplicates(self):
        if self.head is None:   # Jika list kosong, langsung keluar
            return

        seen = set()            # TODO 1: Set untuk menyimpan nilai yang sudah ditemui
        current = self.head     # Mulai dari head
        prev = None             # Node sebelumnya (untuk menghapus node saat ini)

        while current is not None:
            if current.data in seen:
                # TODO 2: Data sudah pernah ada → hapus node ini
                # Sambungkan prev.next langsung ke current.next (lewati current)
                prev.next = current.next
                           # Hapus baris ini setelah mengisi TODO 2
            else:
                # TODO 3: Data belum pernah ada → tambahkan ke seen
                seen.add(current.data)

                # TODO 4: Pindahkan prev ke current (karena current tidak dihapus)
                prev = current
                         # Hapus baris ini setelah mengisi TODO 3 dan 4

            # TODO 5: Maju ke node berikutnya
            current = current.next
```

```python
# ===== MAIN =====
ll = LinkedList()
for val in [1, 3, 2, 3, 5, 1, 2]:
    ll.insert_at_end(val)

print("Sebelum remove_duplicates:")
ll.display()

ll.remove_duplicates()

print("Sesudah remove_duplicates:")
ll.display()

# Output yang diharapkan:
# Sebelum remove_duplicates:
# 1 -> 3 -> 2 -> 3 -> 5 -> 1 -> 2 -> None
# Sesudah remove_duplicates:
# 1 -> 3 -> 2 -> 5 -> None
```
